```python
# Student Grade Calculator

marks = [95, 82, 76, 68, 54, 43, 35, 20]

for mark in marks:
    if mark < 0 or mark > 100:
        grade = "Invalid"
    elif mark >= 90:
        grade = "A+"
    elif mark >= 80:
        grade = "A"
    elif mark >= 70:
        grade = "B"
    elif mark >= 60:
        grade = "C"
    elif mark >= 50:
        grade = "D"
    elif mark >= 40:
        grade = "E"
    else:
        grade = "F"

    print("Marks:", mark, "Grade:", grade)
```
